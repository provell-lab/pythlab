Упражнение 1

```python
import sympy as sp
import numpy as np
from sympy.abc import pi
a = [sp.cos((sp.pi)*k/2) for k in range(1,11)]
for i in range(len(a)):
    k = i + 1
    print(f'k = {k}, значение равно {a[i]}')
print('-------------------')
b = [sp.cos((np.pi)*k/2) for k in range(1,11)]
for j in range(len(a)):
    k = j + 1
    print(f'k = {k}, значение равно {b[j]}')
print('-------------------')
for k in range(len(a)):
    print(a[k],a[k]==b[k],b[k])
```

Упражнение 2

```python
import sympy as sp
from sympy.abc import x

v = (x**7 + 3*x**2 - 4)/(x-1)
print('a)'), sp.pprint(v)
print('После упрощения: ', sp.simplify(v))

c = (x**6 + x**5 + 3*x**4 + 5 * x**3 -6*x**2 + 4*x -8)/(x**2-1)
print('б)'), sp.pprint(c)
print('После упрощения: ', sp.simplify(c))
```

Упражнение 3

```python
import sympy as sp
from sympy.abc import x,y

a = x**4 + 4
b = x**7 + 1
c = 2*x**3 + x**2*y - 2*x*y**2 - y**3

sp.pprint(a)
print('-----------------')
sp.pprint(b)
print('-----------------')
sp.pprint(c)

print('a) После разложения: ', sp.factor(a))
print('b) После разложения: ', sp.factor(b))
print('c) После разложения: ', sp.factor(c))
```

Упражнение 6

```python
import matplotlib.pyplot as plt
import numpy as np
from sympy import *
from sympy.abc import z


n = np.arange(10000, 50000)
n2 = np.arange(1,100)
x = (np.sin(n)+np.cos(n))/(n*(n+1))
plt.plot(n,x,'b.')
plt.title('Последовательность XN')
plt.xlabel('x')
plt.ylabel('xn')
plt.grid()
plt.show()

y = (np.sin(n2)+np.cos(n2))/(n2*(n2+1))
plt.plot(n2,y,'b.')
plt.title('Последовательность XN')
plt.xlabel('x')
plt.ylabel('xn')
plt.grid()
plt.show()


n = np.arange(10000, 50000)
n2 = np.arange(1,100)
x = ((np.log10(10*n)**2))/((np.log10(n)**2))
plt.plot(n,x,'b.')
plt.title('Последовательность XN')
plt.xlabel('x')
plt.ylabel('xn')
plt.grid()
plt.show()

y = ((np.log10(10*n2)**2))/((np.log10(n2)**2))
plt.plot(n2,y,'b.')
plt.title('Последовательность XN')
plt.xlabel('x')
plt.ylabel('xn')
plt.grid()
plt.show()

n = np.arange(10000, 50000)
maxn = n[-1]
rec = 1.0 / np.arange(1, maxn + 1)
H = np.cumsum(rec)
x = H[n - 1]
plt.plot(n, x, 'b.')
plt.title('Последовательность XN')
plt.xlabel('n')
plt.ylabel('H_n')
plt.grid()
plt.show()
```
