```python


import sympy as sp
import numpy as np
from sympy.abc import pi
a = [sp.cos((sp.pi)*k/2) for k in range(1,11)]
for i in range(len(a)):
    k = i + 1
    print(f'k = {k}, значение равно {a[i]}')
print('-------------------')
b = [sp.cos((np.pi)*k/2) for k in range(1,11)]
for j in range(len(a)):
    k = j + 1
    print(f'k = {k}, значение равно {b[j]}')
print('-------------------')
for k in range(len(a)):
    print(a[k],a[k]==b[k],b[k])

```

```python
import sympy as sp
from sympy.abc import x

v = (x**7 + 3*x**2 - 4)/(x-1)
print('a)'), sp.pprint(v)
print('После упрощения: ', sp.simplify(v))

c = (x**6 + x**5 + 3*x**4 + 5 * x**3 -6*x**2 + 4*x -8)/(x**2-1)
print('б)'), sp.pprint(c)
print('После упрощения: ', sp.simplify(c))
```

```python
import sympy as sp
from sympy.abc import x,y

a = x**4 + 4
b = x**7 + 1
c = 2*x**3 + x**2*y - 2*x*y**2 - y**3

sp.pprint(a)
print('-----------------')
sp.pprint(b)
print('-----------------')
sp.pprint(c)

print('a) После разложения: ', sp.factor(a))
print('b) После разложения: ', sp.factor(b))
print('c) После разложения: ', sp.factor(c))
```

Упражнение 6

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.arange(2, 50)

plt.figure(figsize=(15, 9))

# Первый график
plt.subplot(2, 3, 1)
x1 = (np.sin(n) + np.cos(n)) / (n * (n + 1))
plt.plot(n, x1, 'g')
plt.grid()
plt.title('sin и cos')

# Второй график
plt.subplot(2, 3, 2)
x2 = (np.log10(10 * n) ** 2) / (np.log10(n) ** 2)
plt.plot(n, x2, 'b')
plt.grid()
plt.title('log')

# Третий график
plt.subplot(2, 3, 3)
x3 = []
s = 0

for k in n:
    s += 1 / k
    x3.append(s)

x3 = np.array(x3)
plt.plot(n, x3, 'm')
plt.grid(True)
plt.title('Частичные суммы гармонического ряда')


# Во второй части берём меньше точек
n = np.arange(10_000, 50_000, 500)

# Четвёртый график
plt.subplot(2, 3, 4)
x1 = (np.sin(n) + np.cos(n)) / (n * (n + 1))
plt.plot(n, x1, 'g.')
plt.grid()
plt.title('X1 для больших n')
plt.xlabel('n')
plt.ylabel('x1')

# Пятый график
plt.subplot(2, 3, 5)
x2 = (np.log10(10 * n) ** 2) / (np.log10(n) ** 2)
plt.plot(n, x2, 'b.')
plt.grid()
plt.title('X2 для больших n')
plt.xlabel('n')
plt.ylabel('x2')

# Шестой график
plt.subplot(2, 3, 6)
x3 = []
s = 0

for k in range(1, n[-1] + 1):
    s += 1 / k
    if k in n:
        x3.append(s)

plt.plot(n, x3, 'm.')
plt.grid()
plt.title('X3 для больших n')
plt.xlabel('n')
plt.ylabel('H_n')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Последовательность x_n
n = np.arange(1, 51)
x = (2 * n**3 + 2 * n**2 + 3) / (n**3 - 5 * n - 1000)

# График
plt.figure(figsize=(10, 5))
plt.plot(n, x, 'b.')
plt.axhline(y=2, color='r', linestyle='--')
plt.grid(True)
plt.title('Последовательность x_n')
plt.xlabel('n')
plt.ylabel('x_n')
plt.show()

# Предел последовательности
print('Предел последовательности:', 2)

# Поиск n0: для всех n > n0 должно выполняться |x_n - 2| < eps
n = np.arange(1, 100001)
x = (2 * n**3 + 2 * n**2 + 3) / (n**3 - 5 * n - 1000)

n0_1 = np.max(n[np.abs(x - 2) >= 0.01])
n0_2 = np.max(n[np.abs(x - 2) >= 0.001])

print('n0(0.01) =', n0_1)
print('n0(0.001) =', n0_2)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Число сторон многоугольника
n = np.arange(3, 1001)

# Площади многоугольников при R = 2
x = 2 * n * np.sin(2 * np.pi / n)

# График последовательности площадей
plt.figure(figsize=(10, 5))
plt.plot(n, x, 'b.')
plt.axhline(y=4 * np.pi, color='r', linestyle='--')
plt.grid(True)
plt.title('Площади вписанных правильных многоугольников')
plt.xlabel('Количество сторон n')
plt.ylabel('Площадь')
plt.show()

print('Предел последовательности с точностью до тысячных:', round(4 * np.pi, 3))
```
