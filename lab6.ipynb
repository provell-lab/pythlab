Упражнение 1

```python
import sympy as sp
import numpy as np
from sympy.abc import pi
a = [sp.cos((sp.pi)*k/2) for k in range(1,11)]
for i in range(len(a)):
    k = i + 1
    print(f'k = {k}, значение равно {a[i]}')
print('-------------------')
b = [sp.cos((np.pi)*k/2) for k in range(1,11)]
for j in range(len(a)):
    k = j + 1
    print(f'k = {k}, значение равно {b[j]}')
print('-------------------')
for k in range(len(a)):
    print(a[k],a[k]==b[k],b[k])
```

Упражнение 2

```python
import sympy as sp
from sympy.abc import x

v = (x**7 + 3*x**2 - 4)/(x-1)
print('a)'), sp.pprint(v)
print('После упрощения: ', sp.simplify(v))

c = (x**6 + x**5 + 3*x**4 + 5 * x**3 -6*x**2 + 4*x -8)/(x**2-1)
print('б)'), sp.pprint(c)
print('После упрощения: ', sp.simplify(c))
```

Упражнение 3

```python
import sympy as sp
from sympy.abc import x,y

a = x**4 + 4
b = x**7 + 1
c = 2*x**3 + x**2*y - 2*x*y**2 - y**3

sp.pprint(a)
print('-----------------')
sp.pprint(b)
print('-----------------')
sp.pprint(c)

print('a) После разложения: ', sp.factor(a))
print('b) После разложения: ', sp.factor(b))
print('c) После разложения: ', sp.factor(c))
```
