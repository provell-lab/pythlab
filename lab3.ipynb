Упражнение 1

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5,5,100)

y1 = x**2
y2 = 1/x

plt.figure()

plt.plot(x, y1, 'r--o')
plt.title('Парабола y = x^2')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()

plt.figure()
plt.plot(x,y2, 'b-.s')
plt.title('Гипербола y = 1/x')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()

plt.show()
```

Упражнение 2

```python
import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-5,5,100)


plt.figure()
plt.plot(x, np.sin(x), 'b--o',label='y = sin(x)')
plt.plot(x, np.sin(x)-2, 'r--o',label='y = sin(x)-2')
plt.plot(x, np.sin(x)+1, 'g--o',label='y = sin(x)+1')
plt.grid()

plt.xlabel('x')
plt.ylabel('y')
plt.title('Графики функций синуса')
plt.legend()

plt.show()
```

Упражнение 3

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2 * np.pi, 2 * np.pi, 100)

handle1, = plt.plot(x, np.cos(x), 'b-o', label='y = cos(x)')
handle2, = plt.plot(x, 2 * np.cos(x), 'r--s', label='y = 2cos(x)')
handle3, = plt.plot(x, 0.3 * np.cos(x), 'g-.^', label='y = 0.3cos(x)')
handle4, = plt.plot(x, -np.cos(x), 'm:d', label='y = -cos(x)')

plt.legend(handles=[handle1, handle2, handle3, handle4])
plt.xlabel('x')
plt.ylabel('y')
plt.title('Графики функций косинуса')
plt.grid(True)
plt.show()
```

Упражнение 4

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)

def f(x):
    return np.abs(np.abs(x) - 2)

plt.subplot(2, 3, 1)
plt.plot(x, f(x))
plt.title('y = f(x)')
plt.grid()

plt.subplot(2, 3, 2)
plt.plot(x, f(x - 2))
plt.title('y = f(x - 2)')
plt.grid()

plt.subplot(2, 3, 3)
plt.plot(x, f(x + 2))
plt.title('y = f(x + 2)')
plt.grid()

plt.subplot(2, 3, 4)
plt.plot(x, f(2 * x))
plt.title('y = f(2x)')
plt.grid()

plt.subplot(2, 3, 5)
plt.plot(x, f(0.5 * x))
plt.title('y = f(0.5x)')
plt.grid()

plt.subplot(2, 3, 6)
plt.plot(x, f(-x))
plt.title('y = f(-x)')
plt.grid()

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-3, 4, 1)
z = np.ones((7)) * 10
y = np.power(z, x)

plt.subplot(2, 1, 1)
plt.plot(x, y)
plt.grid()

plt.subplot(2, 1, 2)
plt.semilogy(x, y)
plt.grid()

plt.show()
```

Упражнение 7

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, np.pi, 100)
y = np.cos(x)

plt.plot(x, y, 'g:')
plt.plot(y, x, 'r-.')

plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.plot([0, 1.6], [0, 1.6], 'b-')

plt.axis('equal')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Функция и обратная функция')
plt.grid(True)
plt.show()
```

Упражнение 9

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.linspace(-2, 2, 100)
y1 = np.e**(x1)

x2 = np.linspace(np.e**(-2), np.e**(2), 100)
y2 = np.log(x2)

x3 = np.linspace(-2, np.e**(2), 100)
y3 = x3

plt.plot(x1, y1, 'b-o', label='y = e^x')
plt.plot(x2, y2, 'r--s', label='y = ln(x)')
plt.plot(x3, y3, 'g-.^', label='y = x')

plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.show()
```

Упражнение 10

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.1, 5, 100)

plt.subplot(1, 2, 1)
plt.plot(x, x, 'r-o', label='y = x')
plt.plot(x, 1 / x, 'b--s', label='y = 1/x')
plt.plot(x, 1 / x**0.5, 'g-.^', label='y = 1/(x^1/2)')
plt.title('Линейная шкала')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.loglog(x, x, 'r-o', label='y = x')
plt.loglog(x, 1 / x, 'b--s', label='y = 1/x')
plt.loglog(x, 1 / x**0.5, 'g-.^', label='y = 1/(x^1/2)')
plt.title('Логарифмическая шкала')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()

plt.show()
```
