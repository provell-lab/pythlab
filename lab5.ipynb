```python
i = 1j

print('(2 + 3i)(3 - i) =', (2 + 3*i) * (3 - i))
print('-----------')
print('(1 - i)^3 - (1 + i)^3 =', (1 - i)**3 - (1 + i)**3)
print('-----------')
print('i^1 =', i**1)
print('i^2 =', i**2)
print('i^3 =', i**3)
print('i^4 =', i**4)
print('i^5 =', i**5)
print('i^6 =', i**6)
print('i^7 =', i**7)
print('i^8 =', i**8)
print('-----------')
print('(2 - 3i)/(1 + 4i) + 1/(4 - i) =',
      (2 - 3*i) / (1 + 4*i) + 1 / (4 - i))
```

```python
import numpy as np

z = (4 - 5j) * (5 - 6 * (1j**3))

print('z =', z)
print('Сопряженное число:', np.conj(z))
print('Действительная часть:', np.real(z))
print('Мнимая часть:', np.imag(z))
print('Модуль:', np.abs(z))
print('Аргумент:', np.angle(z))

z = (1 + 1j)**15

print('z =', z)
print('Сопряженное число:', np.conj(z))
print('Действительная часть:', np.real(z))
print('Мнимая часть:', np.imag(z))
print('Модуль:', np.abs(z))
print('Аргумент:', np.angle(z))
```

Упражнение 3

```python
import matplotlib.pyplot as plt
import numpy as np

z1 = 1 + 3j
z2 = 3 + 4j
z3 = z1 + z2
z4 = z1 - z2

print('z1 + z2 =', z3)
print('z1 - z2 =', z4)

orig = np.zeros(4)

plt.plot()
plt.quiver(orig, orig,
           np.real([z1, z2, z3, z4]),
           np.imag([z1, z2, z3, z4]),
           angles='xy', scale_units='xy', scale=1,
           color=['blue', 'green', 'red', 'orange'])

plt.axis('equal')
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.grid(True)
plt.title('Комплексные числа и векторы')
plt.xlabel('Re(z)')
plt.ylabel('Im(z)')
plt.show()
```

Упражнение 4

```python
import matplotlib.pyplot as plt
import numpy as np

z = 2 * np.sqrt(3) - 2j
r = np.abs(z)
phi = np.angle(z)

k = np.arange(1, 9)
zroot = r**(1/8) * (
    np.cos((phi + 2.0 * np.pi * k) / 8)
    + 1j * np.sin((phi + 2.0 * np.pi * k) / 8)
)

print(zroot)

plt.plot(np.real(zroot), np.imag(zroot), '*')
plt.plot(np.real(zroot), np.imag(zroot), '-.')

t = np.linspace(0, 2 * np.pi, 100)
plt.plot(r**(1/8) * np.cos(t), r**(1/8) * np.sin(t), ':')

plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.axis('equal')
plt.grid(True)
plt.title('Корни уравнения z^8 = 2√3 - 2i')
plt.xlabel('Re(z)')
plt.ylabel('Im(z)')
plt.show()
```

Упражнение 5

```python
import numpy as np

print('Корни в левой полуплоскости:')
print(zroot[np.real(zroot) < 0])

print('Корни с главным аргументом больше pi/3:')
print(zroot[np.angle(zroot) > np.pi / 3])
```

```python
import matplotlib.pyplot as plt
import numpy as np

r = np.linspace(0, 5, 100)

z1 = r * (np.cos(np.pi / 4) + 1j * np.sin(np.pi / 4))
z2 = r * (np.cos(5 * np.pi / 6) + 1j * np.sin(5 * np.pi / 6))

plt.subplot(1, 2, 1)
plt.plot(np.real(z1), np.imag(z1), 'r')
plt.axis('equal')
plt.grid(True)
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.title('Arg(z) = pi/4')
plt.xlabel('Re(z)')
plt.ylabel('Im(z)')

plt.subplot(1, 2, 2)
plt.plot(np.real(z2), np.imag(z2), 'b')
plt.axis('equal')
plt.grid(True)
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.title('Arg(z) = 5pi/6')
plt.xlabel('Re(z)')
plt.ylabel('Im(z)')

plt.show()
```

```python
import numpy as np
import numpy.random as rng
import matplotlib.pyplot as plt

x = 3 - 6 * rng.rand(10000)
y = 3 - 6 * rng.rand(10000)
z = x + 1j * y

# Условие 1: два сектора с ограничением |z| <= 2
L1 = (np.angle(z) >= np.pi / 6) & (np.abs(z) <= 2)
L2 = (-np.angle(z) >= np.pi / 6) & (np.abs(z) <= 2)

# Круг |z - 0.2 - i| <= 0.2
L3 = np.abs(z - 0.2 - 1j) <= 0.2

# Условие 2: Arg(6i - z) от pi/3 до 2pi/3
L4 = (np.angle(6j - z) >= np.pi / 3) & \
     (np.angle(6j - z) <= 2 * np.pi / 3)

# Прямоугольные области
L5 = (np.abs(np.real(z)) <= 1.7) & \
     (np.imag(z) >= 0) & (np.imag(z) <= 3)

L6 = (np.real(z) >= -1.2) & (np.real(z) <= -0.6) & \
     (np.imag(z) >= 0) & (np.imag(z) <= 2)

L7 = (np.real(z) >= 0.7) & (np.real(z) <= 1.3) & \
     (np.imag(z) >= 1.3) & (np.imag(z) <= 1.9)

# Круги и внешняя область круга
L8 = np.abs(z - 3 - 7j) <= 0.8
L9 = np.abs(z - 5 - 6j) <= 0.5
L10 = np.abs(z - 5.5 - 5.9j) >= 0.6

plt.plot(x[L1], y[L1], '.', color='blue')
plt.plot(x[L2], y[L2], '.', color='red')
plt.plot(x[L3], y[L3], '.', color='green')

plt.plot(x[L4], y[L4], '.', color='orange')
plt.plot(x[L5], y[L5], '.', color='purple')
plt.plot(x[L6], y[L6], '.', color='brown')
plt.plot(x[L7], y[L7], '.', color='black')
plt.plot(x[L8], y[L8], '.', color='cyan')
plt.plot(x[L9], y[L9], '.', color='magenta')
plt.plot(x[L10], y[L10], '.', color='gray')

plt.axis('equal')
plt.grid(True)
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.title('Множества точек на комплексной плоскости')
plt.xlabel('Re(z)')
plt.ylabel('Im(z)')
plt.show()
```
