Упражнение 1

```python
import sympy as sp
from sympy.abc import x

f = x**(1/3)
f2 = 2**x
def pr(f,x1):
    p = float(sp.diff(f,x).subs(x,x1))
    return p

print(pr(f,1))
print(pr(f,2))
print(pr(f,3))
print('-------------------------')
print(pr(f2,1))
print(pr(f2,2))
print(pr(f2,3))
```

Уапражнение 2

```python
import sympy as sp
from sympy.abc import x

f = sp.atan(x**(1/2))**2
f2 = 3**(sp.asin(x**2))
def pr(f,x1):
    p = float(sp.diff(f,x).subs(x,x1))
    return p


print(pr(f,0.5))
print('-------------------------')
print(pr(f2,0.5))
```

Упражнение 3

```python
import matplotlib.pyplot as plt
import sympy as sp
import numpy as np
from sympy.abc import x

f = x**(1/3)
x0 = 1
def graph(f,x0):
    n = np.arange(-10, 10, 0.1)
    ff = sp.lambdify(x, f, modules='numpy')
    fn = sp.lambdify(x, sp.diff(f, x), modules='numpy')
    y = ff(n)
    dy = fn(n)
    plt.plot(n, y, label=f'f(x) = {str(f)}', color='blue')
    plt.plot(n, dy, label=f"f'(x) = {sp.diff(f,x)}", color='red', linestyle='--')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Функция и её производная')
    plt.legend()
    plt.grid()
    plt.show()

graph(f,x0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

T = np.arange(0, 801, 10)

Q = 0.3965 * T + 2.081 * 10**-3 * T**2 - 5.024 * 10**-7 * T**3
c = 0.3965 + 4.162 * 10**-3 * T - 1.5072 * 10**-6 * T**2

plt.subplot(2, 1, 1)
plt.plot(T, Q)
plt.xlabel("Температура T, °C")
plt.ylabel("Q(T), Дж")
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(T, c)
plt.xlabel("Температура T, °C")
plt.ylabel("c(T), Дж/°C")
plt.grid()

plt.tight_layout()
plt.show()
```
