Упражнение 1

```python
import matplotlib.pyplot as plt
import numpy as np

A = np.array([-2, 0])
B = np.array([1, 2])
C = np.array([1, -1])

AB = B - A
BC = C - B
CD = B - C

D = A + C - B
CD = D - C
DA = A - D

print('Координаты вершины D:', D)
print('Вектор AB:', AB)
print('Вектор BC:', BC)
print('Вектор CD:', CD)
print('Вектор DA:', DA)

plt.grid()
plt.xlim(-5, 5)
plt.ylim(-5, 5)

plt.arrow(A[0], A[1], AB[0], AB[1], width=0.03, length_includes_head=True, color='blue')
plt.arrow(B[0], B[1], BC[0], BC[1], width=0.03, length_includes_head=True, color='red')
plt.arrow(C[0], C[1], CD[0], CD[1], width=0.03, length_includes_head=True, color='green')
plt.arrow(D[0], D[1], DA[0], DA[1], width=0.03, length_includes_head=True, color='orange')

plt.text(A[0] - 0.3, A[1] + 0.1, 'A')
plt.text(B[0] + 0.1, B[1] + 0.1, 'B')
plt.text(C[0] + 0.1, C[1] + 0.1, 'C')
plt.text(D[0] - 0.3, D[1] + 0.1, 'D')

plt.show()
```

Упражнение 2

```python
import matplotlib.pyplot as plt
import numpy as np

A = np.array([-2, 0])
B = np.array([3, 4])
C = np.array([4, -1])

AB = B - A
AC = C - A
BA = A - B
BC = C - B
CA = A - C
CB = B - C

cos_A = np.dot(AB, AC) / np.linalg.norm(AB) / np.linalg.norm(AC)
cos_B = np.dot(BA, BC) / np.linalg.norm(BA) / np.linalg.norm(BC)
cos_C = np.dot(CA, CB) / np.linalg.norm(CA) / np.linalg.norm(CB)

angle_A = np.arccos(cos_A) * 180 / np.pi
angle_B = np.arccos(cos_B) * 180 / np.pi
angle_C = np.arccos(cos_C) * 180 / np.pi

print('Угол A:', angle_A)
print('Угол B:', angle_B)
print('Угол C:', angle_C)
print('Сумма углов:', angle_A + angle_B + angle_C)

plt.plot([A[0], B[0]], [A[1], B[1]], 'b-')
plt.plot([B[0], C[0]], [B[1], C[1]], 'r-')
plt.plot([C[0], A[0]], [C[1], A[1]], 'g-')

plt.plot(A[0], A[1], 'ko')
plt.text(A[0] + 0.1, A[1] + 0.1, 'A')

plt.plot(B[0], B[1], 'ko')
plt.text(B[0] + 0.1, B[1] + 0.1, 'B')

plt.plot(C[0], C[1], 'ko')
plt.text(C[0] + 0.1, C[1] + 0.1, 'C')

plt.grid()
plt.show()
```

Упражнение 4

```python
import matplotlib.pyplot as plt
import numpy as np

ax = plt.axes(projection='3d')

z = np.array([0, 0, 0])
a = np.array([1, -2, 0])
b = np.array([0, 1, 1])
c = np.array([1, 2, 2])

rez = np.dot(np.cross(a, b), c)

print('Смешанное произведение:', rez)

if rez != 0:
    print('Векторы образуют базис.')
else:
    print('Векторы не образуют базис.')

ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_zlim([-3, 3])

ax.quiver(*z, *a, color='Blue')
ax.quiver(*z, *b, color='Green')
ax.quiver(*z, *c, color='Red')

plt.grid()
plt.show()
```

Упражнение 5

```python
import matplotlib.pyplot as plt
import numpy as np

ax = plt.axes(projection='3d')

A = np.array([1, 3, -1])
B = np.array([2, -1, 4])
C = np.array([5, 0, 3])

AB = B - A
AC = C - A
n = np.cross(AB, AC)

print('Вектор AB:', AB)
print('Вектор AC:', AC)
print('Векторное произведение AB и AC:', n)
print('Площадь треугольника:', np.linalg.norm(n) / 2)

ax.plot([A[0], B[0]], [A[1], B[1]], [A[2], B[2]], color='Blue')
ax.plot([B[0], C[0]], [B[1], C[1]], [B[2], C[2]], color='Green')
ax.plot([C[0], A[0]], [C[1], A[1]], [C[2], A[2]], color='Red')

ax.text(A[0], A[1], A[2], 'A')
ax.text(B[0], B[1], B[2], 'B')
ax.text(C[0], C[1], C[2], 'C')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.grid()
plt.show()
```

Упражнение 6

```python
import matplotlib.pyplot as plt
import numpy as np

ax = plt.axes(projection='3d')

A = np.array([0, 0, 0])
AB = np.array([3, 4, 0])
AC = np.array([-3, 0, 1])
AD = np.array([0, 2, 5])

B = A + AB
C = A + AC
D = A + AD

volume = abs(np.dot(AB, np.cross(AC, AD))) / 6

print('Объём пирамиды ABCD:', volume)

ax.plot([A[0], B[0]], [A[1], B[1]], [A[2], B[2]], color='Blue')
ax.plot([A[0], C[0]], [A[1], C[1]], [A[2], C[2]], color='Green')
ax.plot([A[0], D[0]], [A[1], D[1]], [A[2], D[2]], color='Red')
ax.plot([B[0], C[0]], [B[1], C[1]], [B[2], C[2]], color='Blue')
ax.plot([B[0], D[0]], [B[1], D[1]], [B[2], D[2]], color='Green')
ax.plot([C[0], D[0]], [C[1], D[1]], [C[2], D[2]], color='Red')

ax.text(A[0], A[1], A[2], 'A')
ax.text(B[0], B[1], B[2], 'B')
ax.text(C[0], C[1], C[2], 'C')
ax.text(D[0], D[1], D[2], 'D')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.grid()
plt.show()
```
