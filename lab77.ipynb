```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np
t = Symbol('t')

# a)
x = np.linspace(0.01,7,1000)
a = (e**(4*x) - 1)/(1-e**(5*x))
a1 = (e**(4*t) - 1)/(1-e**(5*t))


plt.plot(x, a,'b')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)
plt.title('График при x-> +oo')
plt.show()

print(f'Лимит a) равен: {limit(a1,t,oo)}')

# б)
x = np.linspace(0.8, 1.2, 1000)
f = (-x**2-3*x-7)*np.exp(2*x)
f1 = (-t**2-3*t-7)*sp.exp(2*t)

plt.plot(x, f,'b')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.title('График в окрестонсти x = 1 ')
plt.show()

print(f'Лимит б) равен: {limit(f1,t,1)}')

# в)
x = np.linspace(-0.1, 0.1, 1000)
fn = (np.sin(x) / (np.sqrt(x + 1) - np.sqrt(2*x + 1)))
ft = (sp.sin(t))/((t+1)**1/2 - (2*t + 1)**1/2)
v = limit(ft,t,0)

plt.plot(x, fn,'b')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.title('График в окрестонсти x = 0 ')
plt.show()

print(f'Лимит в) равен: {limit(ft,t,0)}')
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

#a)

x1 = np.linspace(-2,-0.1,1000)
x2 = np.linspace(0.1,2,1000)
f1 = (abs(x1)-x1)/x1**2
f2 = (abs(x2)-x2)/x2**2
plt.subplot(2,2,1)
plt.plot(x1,f1,'b')
plt.plot(x2,f2,'b')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.title('a) Разрыв при x = 0')


#б)

plt.subplot(2,2,2)
plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)
x1 = np.linspace(-2,-0.01,1000)
x2 = np.linspace(0.01,2,1000)
f1 = (x1+1)/np.atan(1/x1)
f2 = (x2+1)/np.atan(1/x2)
plt.plot(x1,f1,'b')
plt.plot(x2,f2,'b')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.title('б) Скачок при x = 0')

#в)

plt.subplot(2,2,3)
plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)
x1 = np.linspace(-2,-0.01,1000)
x2 = np.linspace(0.01,2,1000)
f1 = np.sin(x1)*np.sin(1/x1)
f2 = np.sin(x2)*np.sin(1/x2)
plt.plot(x1,f1,'b')
plt.plot(x2,f2,'b')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.title('в) Устранимый разрыв при x = 0')


#г)

plt.subplot(2,2,4)
plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)
x1 = np.linspace(-2,-0.01,1000)
x2 = np.linspace(0.01,0.99,1000)
x3 = np.linspace(1.01, 2,1000)
f1 = np.cos((np.pi*x1/2))/(x1**3-x1**2)
f2 = np.cos((np.pi*x2/2))/(x2**3-x2**2)
f3 = np.cos((np.pi*x3/2))/(x3**3-x3**2)
plt.plot(x1,f1,'b')
plt.plot(x2,f2,'b')
plt.plot(x3,f3,'b')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.title('г) Разрывы при x = 0')
plt.xlim(-2, 2)
plt.ylim(-10, 3)

plt.tight_layout()
```

Упражнение 3

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

t = sp.Symbol('t')
plt.figure(figsize=(10, 9))

# а)

f = (4*t**3 + 5*t**2 - 1)/(t**2 - 9)

print('а)')
print('Предел при x -> -3 слева:', sp.limit(f, t, -3, dir='-'))
print('Предел при x -> -3 справа:', sp.limit(f, t, -3, dir='+'))
print('Предел при x -> 3 слева:', sp.limit(f, t, 3, dir='-'))
print('Предел при x -> 3 справа:', sp.limit(f, t, 3, dir='+'))

k1 = sp.limit(f/t, t, -sp.oo)
b1 = sp.limit(f - k1*t, t, -sp.oo)
k2 = sp.limit(f/t, t, sp.oo)
b2 = sp.limit(f - k2*t, t, sp.oo)

print(f'При x → -oo: k = {k1}, b = {b1}')
print(f'При x → +oo: k = {k2}, b = {b2}')
print('Асимптоты: x = -3, x = 3, y = 4x + 5')

plt.subplot(2, 1, 1)
plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)

x1 = np.linspace(-10, -3.01, 1000)
x2 = np.linspace(-2.99, 2.99, 1000)
x3 = np.linspace(3.01, 10, 1000)

f1 = (4*x1**3 + 5*x1**2 - 1)/(x1**2 - 9)
f2 = (4*x2**3 + 5*x2**2 - 1)/(x2**2 - 9)
f3 = (4*x3**3 + 5*x3**2 - 1)/(x3**2 - 9)

plt.plot(x1, f1, 'b')
plt.plot(x2, f2, 'b')
plt.plot(x3, f3, 'b')

x = np.linspace(-10, 10, 1000)
plt.plot(x, 4*x + 5, 'r--')
plt.title('y = 4x + 5')
plt.axvline(x=-3, color='green', linestyle='--', label='x = -3')
plt.axvline(x=3, color='orange', linestyle='--', label='x = 3')

plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.title('а) Асимптоты рациональной функции')
plt.xlim(-10, 10)
plt.ylim(-60, 60)

# б)

f = t*sp.exp(t)

k = sp.limit(f/t, t, -sp.oo)
b = sp.limit(f - k*t, t, -sp.oo)

print('б)')
print(f'При x -> -oo: k = {k}, b = {b}')
print('Горизонтальная асимптота: y = 0 при x -> -oo')
print('При x -> +oo предел f(x)/x:', sp.limit(f/t, t, sp.oo))

plt.subplot(2, 1, 2)
plt.axvline(x=0, color='black', linewidth=1)

x = np.linspace(-8, 2, 1000)
f = x*np.exp(x)

plt.plot(x, f, 'b')
plt.axhline(y=0, color='red', linestyle='--', label='y = 0')

plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.title('б) Горизонтальная асимптота при x -> -oo')
plt.xlim(-8, 2)
plt.ylim(-1, 16)

plt.tight_layout()
plt.show()
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

t = sp.Symbol('t')

f1 = sp.sinh(t)/t - 1
f2 = t*sp.log(t**2)

left = sp.limit(f1, t, 0, dir='-')
right = sp.limit(f2, t, 0, dir='+')

print(f'Предел слева: {left}')
print(f'Предел справа: {right}')
print(f'Для непрерывности a = {left}')

x1 = np.linspace(-1, -0.001, 1000)
x2 = np.linspace(0.001, 1, 1000)

f1 = np.sinh(x1)/x1 - 1
f2 = x2*np.log(x2**2)

plt.figure(figsize=(8, 5))
plt.plot(x1, f1, 'b')
plt.plot(x2, f2, 'b')
plt.plot(0, 0, 'ro', label='a = 0')

plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.title('4) Непрерывность при a = 0')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Между бусинками увеличивается только масса нити
x1 = np.linspace(0, 1, 1000)
x2 = np.linspace(1, 2, 1000)
x3 = np.linspace(2, 3, 1000)
x4 = np.linspace(3, 4, 1000)
x5 = np.linspace(4, 5, 1000)
x6 = np.linspace(5, 6, 1000)

f1 = 0.039*x1 + 0.3
f2 = 0.039*x2 + 0.6
f3 = 0.039*x3 + 0.9
f4 = 0.039*x4 + 1.2
f5 = 0.039*x5 + 1.5
f6 = 0.039*x6 + 1.8

plt.figure(figsize=(10, 5))

plt.plot(x1, f1, 'b')
plt.plot(x2, f2, 'b')
plt.plot(x3, f3, 'b')
plt.plot(x4, f4, 'b')
plt.plot(x5, f5, 'b')
plt.plot(x6, f6, 'b')

# Выколотые точки: значения перед добавлением бусинки
x = np.arange(1, 7)
y = 0.039*x + 0.3*x
plt.plot(x, y, 'bo', markerfacecolor='white', zorder=3)

# Закрашенные точки: значения с учётом новой бусинки
x = np.arange(0, 7)
y = 0.039*x + 0.3*(x + 1)
plt.plot(x, y, 'bo', zorder=4)

plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)
plt.xlabel('x, см')
plt.ylabel('M(x), г')
plt.grid()
plt.title('5) Масса нити с бусинками')
plt.xlim(-0.2, 6.2)
plt.ylim(0, 2.5)

plt.tight_layout()
plt.show()
```
